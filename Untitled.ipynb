```python
def make_logger(target):
    def logger(data):
        with open(target, 'a') as f:
            f.write(data + '\n')
    return logger

foo_logger = make_logger('foo.txt') #foo.txt will be created if not there already
foo_logger('Hello')
foo_logger('World')
```

```python
foo_logger = make_logger("foo.txt")
bar_logger = make_logger("bar.txt")

foo_logger("run started")
bar_logger("different experiment")
```

```python
def fib(x):
    if x==0 or x==1:
        return 1
    else:
        return fib(x-1)+fib(x-2)
    
```

```python
for i in range(0,2): print(fib(i))
    
```

```python

def fib(x):
    a, b = 0, 1
    for i in range(1,x+1):
        a, b = b, a+b
    return b
        
    
```

```python
print([fib(i) for i in range(5)])
```
